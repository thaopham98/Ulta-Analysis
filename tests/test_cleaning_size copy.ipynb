````python
def parse_size(size_str):
    """
    Parse size strings like "2.5 ml" into (value, unit, status)
    
    Returns:
        tuple: (float_value, unit_string, status_string)
        - status: 'ok', 'missing', 'no_match', 'multiple_units'
    """
    # Handle missing/null values
    if pd.isna(size_str) or not isinstance(size_str, str):
        return None, None, 'missing'
    
    size_str = size_str.strip()
    pattern = r'(\d+\.?\d*)\s*([a-zA-Z]+\.?)'
    
    # Find ALL matches to detect errors
    matches = re.findall(pattern, size_str)
    
    if len(matches) == 0:
        # Pattern didn't match at all
        return None, None, 'no_match'
    elif len(matches) == 1:
        # Perfect: exactly one value-unit pair
        value, unit = matches[0]
        return float(value), unit.rstrip('.'), 'ok'
    else:
        # ERROR: multiple value-unit pairs found (like "0.17 oz. oz")
        value, unit = matches[0]
        return float(value), unit.rstrip('.'), 'multiple_units'

# Apply to dataframe
df_parsed = single_color_swatch_image.copy()
df_parsed[['value', 'unit', 'status']] = df_parsed['size_text'].apply(
    lambda x: pd.Series(parse_size(x))
)

# Analyze results
print("\n=== Status Summary ===")
print(df_parsed['status'].value_counts())

print("\n=== Units (Valid Only) ===")
valid = df_parsed[df_parsed['status'] == 'ok']
print(valid['unit'].value_counts())

print("\n=== Errors ===")
errors = df_parsed[df_parsed['status'] != 'ok']
print(errors[['size_text', 'status']])
````

### Fill in Missing `size_text`

- Merge `product_name` into the a dataframe
- create a reverse lookup from `SIZE_MAP_METRIC`
- Filter and fill missing `size_text`
- Clean up

````python
df1 = single_color_swatch_image.merge(
    products[['product_id', 'product_name']].drop_duplicates(), # avoid create duplicated rows since `product_id`` appears multiple times in `single_color_swatch_image`
    on = 'product_id', # merge on `product_id`
    how='left', # left join (single_color_swatch_image)
    validate='many_to_one' 
)
df1
````

````python
## Invert the dict to lookup size_text by product_name
product_to_size = {}
for size, product_names in SIZE_MAP_METRIC.items():
    for product_name in product_names:
        product_to_size[product_name] = size

product_to_size
````

````python
## Filter and fill missing size_text
## Only update rows where size_text is missing AND the product_name exists 

# Create a mask for rows that need update
mask = (
    df1['size_text'].isna() # `size_text` is missing
    &
    df1['product_name'].isin(product_to_size) # `product_name` exists in `product_to_size`
)

df1.loc[mask, 'size_text'] = df1.loc[mask, 'product_name'].map(product_to_size)
````

````python
# Get missing product_ids
missing_mask3 = df3['size_text'].isna() | df3['size_text'].astype("string").str.strip().eq("")
missing_ids = df3.loc[missing_mask3, 'product_id'].unique()
````

````python
# Create a boolean mask for each missing product_id found in shade_url
def find_product_id_in_url(url, product_id):
    return product_id in str(url)


# Build lookup: product_id -> size
size_lookup = {}
for pid in missing_ids:
    mask = df2['shade_url'].apply(lambda url: pid in str(url))
    matching_rows = df2[mask]
    if not matching_rows.empty:
        size_val = matching_rows['size'].dropna().astype(str).str.strip().replace('', np.nan).dropna()
        if not size_val.empty:
            size_lookup[pid] = size_val.iloc[0]

# Fill missing size_text
df3.loc[missing_mask3, 'size_text'] = df3.loc[missing_mask3, 'product_id'].map(size_lookup)
````

````python
## conditions: df1['product_id'] == df3['product_id'] & df3['size_text'] != None

## Step 1: Create a lookup dictionary from df3
### This dict maps product_id to size_text (skipping NaN values)
lookup = df3.set_index('product_id')['size_text'].dropna().to_dict() # using product_id as the key, size_text as the value

## Step 2: Use map() to fill missing values (.fillna()) in df1
df1['size_text'] = df1['size_text'].fillna(df1['product_id'].map(lookup))
````
