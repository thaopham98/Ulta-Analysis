````python
def parse_size(row):
    match = re.fullmatch(
        r"\s*(\d*\.?\d+)\s*(.*?)\s*",
        str(row["size_text"]),
    )

    if not match:
        return row["sku_id"], None, None

    value = float(match.group(1))
    unit = match.group(2).strip()

    return row["sku_id"], value, unit


parsed_sizes = single_color_swatch_image.apply(
    parse_size,
    axis=1,
    result_type="expand",
)

parsed_sizes.columns = ["sku_id", "value", "unit"]
````

````python
def parse_size(size_str):
    """
    Parse size strings like "2.5 ml" into (value, unit, status)
    
    Returns:
        tuple: (float_value, unit_string, status_string)
        - status: 'ok', 'missing', 'no_match', 'multiple_units'
    """
    # Handle missing/null values
    if pd.isna(size_str) or not isinstance(size_str, str):
        return None, None, 'missing'
    
    size_str = size_str.strip()
    pattern = r'(\d+\.?\d*)\s*([a-zA-Z]+\.?)'
    
    # Find ALL matches to detect errors
    matches = re.findall(pattern, size_str)
    
    if len(matches) == 0:
        # Pattern didn't match at all
        return None, None, 'no_match'
    elif len(matches) == 1:
        # Perfect: exactly one value-unit pair
        value, unit = matches[0]
        return float(value), unit.rstrip('.'), 'ok'
    else:
        # ERROR: multiple value-unit pairs found (like "0.17 oz. oz")
        value, unit = matches[0]
        return float(value), unit.rstrip('.'), 'multiple_units'
````

````python
DUPLICATE_UNIT_PATTERN = re.compile(
    r"\b([a-zA-Z]+)\.?\s+\1\.?\s*$",
    flags=re.IGNORECASE,
)

def parse_size(size_text):
    if pd.isna(size_text):
        return None, None

    cleaned_text = str(size_text).strip()

    cleaned_text = DUPLICATE_UNIT_PATTERN.sub(
        lambda match: match.group(1).lower(),
        cleaned_text
    )

    match = re.fullmatch(
        r"\s*(\d*\.?\d+)\s*(.*?)\s*",
        cleaned_text,
    )

    if not match:
        return None, None

    value = float(match.group(1))
    unit = match.group(2).strip()

    return cleaned_text, value, unit
````
