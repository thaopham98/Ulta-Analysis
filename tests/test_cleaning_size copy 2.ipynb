````python
def parse_size(size_str):
    """
    Parse size strings like "2.5 ml" into (value, unit, status)
    
    Returns:
        tuple: (float_value, unit_string, status_string)
        - status: 'ok', 'missing', 'no_match', 'multiple_units'
    """
    # Handle missing/null values
    if pd.isna(size_str) or not isinstance(size_str, str):
        return None, None, 'missing'
    
    size_str = size_str.strip()
    pattern = r'(\d+\.?\d*)\s*([a-zA-Z]+\.?)'
    
    # Find ALL matches to detect errors
    matches = re.findall(pattern, size_str)
    
    if len(matches) == 0:
        # Pattern didn't match at all
        return None, None, 'no_match'
    elif len(matches) == 1:
        # Perfect: exactly one value-unit pair
        value, unit = matches[0]
        return float(value), unit.rstrip('.'), 'ok'
    else:
        # ERROR: multiple value-unit pairs found (like "0.17 oz. oz")
        value, unit = matches[0]
        return float(value), unit.rstrip('.'), 'multiple_units'

# Apply to dataframe
df_parsed = multi_color_swatch_image.copy()
df_parsed[['value', 'unit', 'status']] = df_parsed['size_text'].apply(
    lambda x: pd.Series(parse_size(x))
)

# Analyze results
print("\n=== Status Summary ===")
print(df_parsed['status'].value_counts())

print("\n=== Units (Valid Only) ===")
valid = df_parsed[df_parsed['status'] == 'ok']
print(valid['unit'].value_counts())

print("\n=== Errors ===")
errors = df_parsed[df_parsed['status'] != 'ok']
print(errors[['size_text', 'status']])
````

### Merge Metadata & Sanitize Text Strings

Merging product metadata (`products`) with variant dataframe (`single_color_swatch_image`) and clean hidden characters like the non-breaking space `\u00a0`.

````python
# Merge product name into variant dataframe for lookup capabilities
df = multi_color_swatch_image.merge(
    products[['product_id','brand','product_name']], # add these columns of products
    on = 'product_id',
    how = 'left',
    validate = 'many_to_one'
)
````

````python
# Clean non-breaking spaces and whitespace across all string columns
string_cols = df.select_dtypes(include=['object', 'string']).columns # columns that is str or object dtype

# print(string_cols)

## Strip
for col in string_cols:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace('\u00a0', ' ', regex=False)
        .str.strip()
    ) # find

    # Convert str literal 'nan' back to actual null values
    df[col] = df[col].replace({'nan': np.nan, '': np.nan})
````

### Clean Prefixes
Convert non-size into NaN so imputation can replace them later

````python
df['size_text_clean'] = df['size_text'].copy()

df.loc[df['size_text_clean'].str.contains(r'(?i)^color:', na=False), 'size_text_clean'] = np.nan
````

````python
def find_size(product_id):
    if pd.isna(product_id):
        return pd.NA

    matching_sizes = (
        df2.loc[
            df2['shade_url'].astype('string').str.contains(
                str(product_id),
                regex=False,
                na=False,
            ),
            "size",
        ]
        .dropna()
        .astype(str)
        .str.strip()
    )
    matching_sizes = matching_sizes[matching_sizes.ne("")]
    if matching_sizes.empty:
        return pd.NA
    return matching_sizes.iloc[0]
````

````python
# 2. Fix salvageable typos on remaining legitimate size strings (e.g., "0.12 oz. oz")
df['size_text_clean'] = (
    df['size_text_clean']
    .str.replace(r'(?i)\boz\.?\s*oz\.?\b', 'oz', regex=True)
    .str.replace(r'(?i)\bg\.?\s*g\.?\b', 'g', regex=True)
    .str.replace(r'(?i)\bfl\.?\s*oz\.?\b', 'fl oz', regex=True)
    .str.replace(r'(?i)\boz\.\b', 'oz', regex=True)
)
````

### Hiearchical Missing Value Imputation

Constructing a fallback hierarchy to fill the **missing `size_text` entries**:

- **Sibling Imputation**: Fill missing sizes within the same `product_id` (variants of the same product almost always share the exact same volume).

- **Reverse Product Name Map**: Turn your `SIZE_MAP_OZ` into a dictionary lookup mapping `product_name` -> `size_text`.

- **Product ID Overrides**: Use your manual `map` dictionary for edge cases (e.g., `"2 ct"` conversions).

````python
# 1. Reverse-map your SIZE_MAP_OZ dictionary to map product_name -> size_text
name_to_size_map = {} 
for size, name_list in SIZE_MAP_OZ.items():
    for name in name_list:
        name_to_size_map[name] = size # keys are product_name, values are size_text 
````

````python
# 2. Level 1: Impute from other SKUs sharing the same product_id
df['size_text_clean'] = df.groupby('product_id')['size_text_clean'].transform(lambda x: x.ffill().bfill())
````

````python
# 3. Level 2: Fallback to product_name lookup via SIZE_MAP_OZ
df['size_text_clean'] = df['size_text_clean'].fillna(df['product_name'].map(name_to_size_map))
````

````python
# 4. Level 2: Fallback to product_id lookup via df2['shade_url']
df['size_text_clean'] = df['size_text_clean'].fillna(df['product_id'].map(find_size))
````

````python
# 4. Level 3: Manual product_id override map
# "mkt77008719": "0.4 oz"  # total volume for 2 ct item

# Forcefully overwrite the size for this specific product ID
df.loc[df['product_id'] == 'mkt77008719', 'size_text_clean'] = '0.4 oz'
df.loc[df['product_id']=="mkt77008719", ['product_id', 'size_text', 'size_text_clean']]
````

### Vectorized Regex Size Extraction

````python
# Regex captures floating/integer numbers and the trailing unit string
pattern = r'^\s*(\d*\.?\d+)\s*([a-zA-Z]+|\S+)?'

extracted = df['size_text_clean'].str.extract(pattern)
df['size_value'] = pd.to_numeric(extracted[0], errors='coerce')
df['size_unit'] = extracted[1].str.lower().str.strip()
````

````python
# Normalize common unit synonyms
unit_mapping = {
    'g': 'g',
    'gram': 'g',
    'grams': 'g',
    'oz': 'oz',
    'fl oz': 'fl oz',
    'ml': 'ml',
}
df['size_unit'] = df['size_unit'].map(unit_mapping).fillna(df['size_unit'])
````

### Unit Standardization & Feature Engineering

````python
# Standardize conversions (1 oz ~ 28.3495 g)
OZ_TO_G = 28.3495
````

````python
def convert_to_grams_ml(row):
    val = row['size_value']
    unit = row['size_unit']
    
    if pd.isna(val) or pd.isna(unit):
        return np.nan
    if unit in ['g', 'ml']:
        return val
    elif unit in ['oz', 'fl oz']:
        return round(val * OZ_TO_G, 2)
    return np.nan
````

````python
# df_parsed = df.drop(columns=['variant_type', 'size_text', 'availability'])


df_parsed['normalized_size_g_ml'] = df_parsed.apply(convert_to_grams_ml, axis=1) # calling convert_to_grams_ml()

# # Calculate Price Per Unit metrics
df_parsed['effective_price'] = df_parsed['sale_price'].fillna(df_parsed['list_price']) # replace list_price with sale_price when sale_price is available
df_parsed['price_per_g_ml'] = (df_parsed['effective_price'] / df_parsed['normalized_size_g_ml']).round(2)
````
